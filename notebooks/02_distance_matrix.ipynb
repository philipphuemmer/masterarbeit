# 02 – Distanzmatrix
Fahrzeit-Matrix über Google Maps Distance Matrix API oder Haversine-Fallback.

```python
import sys
sys.path.insert(0, '..')

from src.data.loader import load_config, load_stations, get_coordinates
from src.api.google_maps import build_haversine_matrix, fetch_distance_matrix
from src.utils.visualization import plot_distance_matrix

config = load_config('../configs/config.yaml')
config['data']['raw_path'] = '../data/raw/charging_stations_wue.csv'

df = load_stations(config)
coords = get_coordinates(df, config)
print(f'Koordinaten (inkl. Depot): {len(coords)}')
```

## Option A: Haversine-Schätzung (kein API-Key nötig)
Gut für Entwicklung und Tests.

```python
dur_h, dist_h = build_haversine_matrix(
    coords,
    avg_speed_kmh=30.0,
    cache_path='../data/distance_matrices/haversine',
)
print('Fahrzeit-Matrix (Haversine) Shape:', dur_h.shape)
print(f'Mittlere Fahrzeit: {dur_h.mean()/60:.1f} min')
plot_distance_matrix(dur_h / 60, 'Fahrzeit-Matrix Haversine [min]')
```
